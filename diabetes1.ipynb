```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('diabetes.csv')
```

```python
df.head()
```

```python
zero_columns = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for col in zero_columns:
    df[col] = df[col].replace(0, np.nan)

print(df.isnull().sum())
```

```python
df.head()
```

```python
for col in zero_columns:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
for col in zero_columns:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
for col in zero_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.head()
```

```python
X = df.drop(columns=['Outcome'])
```

```python
y = df['Outcome']
```

```python
X
```

```python
y
```

```python
from sklearn.preprocessing  import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[0])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
print(f"Training දත්ත ප්‍රමාණය: {X_train.shape[0]}")
print(f"Testing දත්ත ප්‍රමාණය: {X_test.shape[0]}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Model එක සාදා පුහුණු කිරීම
log_model = LogisticRegression()
log_model.fit(X_train, y_train)

# අනාවැකි ලබා ගැනීම
y_pred_log = log_model.predict(X_test)

# Accuracy එක බැලීම
log_acc = accuracy_score(y_test, y_pred_log)
print(f"Logistic Regression Accuracy: {log_acc * 100:.2f}%")
print("\nDetailed Classification Report:\n", classification_report(y_test, y_pred_log))
```

```python
from xgboost import XGBClassifier

# Model එක සාදා පුහුණු කිරීම
xgb_model = XGBClassifier(random_state=42, eval_metric='logloss')
xgb_model.fit(X_train, y_train)

# අනාවැකි ලබා ගැනීම
y_pred_xgb = xgb_model.predict(X_test)

# Accuracy එක බැලීම
xgb_acc = accuracy_score(y_test, y_pred_xgb)
print(f"XGBoost Accuracy: {xgb_acc * 100:.2f}%")
print("\nDetailed Classification Report:\n", classification_report(y_test, y_pred_xgb))
```

```python
def get_diabetes_recommendations(patient_data):
    # 1. Model එකෙන් දියවැඩියාව ඇතිවීමේ සම්භාවිතාව (Probability) ලබා ගැනීම
    # [0][1] මගින් Class 1 (දියවැඩියාව ඇතිවීම) සඳහා සම්භාවිතාව ලබාදෙයි
    risk_prob = log_model.predict_proba(patient_data)[0][1]
    risk_percentage = risk_prob * 100
    
    print(f"📊 ඔබේ දියවැඩියා අවදානම් මට්ටම (Diabetes Risk Rate): {risk_percentage:.2f}%\n")
    print("--------------------------------------------------")
    print("💡 වෛද්‍ය සහ ජීවන රටා උපදෙස් (Recommendations):")
    print("--------------------------------------------------")
    
    # 2. අවදානම අනුව උපදෙස් කාණ්ඩ කිරීම (Rule-based Logic)
    if risk_percentage < 30:
        print("✅ සුභ පැතුම්! ඔබේ අවදානම ඉතා අවම මට්ටමක පවතී.")
        print("* දැනට පවත්වාගෙන යන සෞඛ්‍ය සම්පන්න ජීවන රටාව ඉදිරියටත් රැකගන්න.")
        print("* දිනපතා ප්‍රමාණවත් පරිදි ජලය පානය කරන්න සහ ක්‍රියාශීලී වන්න.")
        
    elif 30 <= risk_percentage < 70:
        print("⚠️ මධ්‍යස්ථ අවදානමක් (Moderate Risk)! ප්‍රවේශම් විය යුතුයි.")
        print("* ආහාර වේලෙහි අඩංගු පැණිරස කෑම සහ කාබෝහයිඩ්‍රේට් (බත්, පාන් වැනි දෑ) සීමා කරන්න.")
        print("* දිනකට අවම වශයෙන් විනාඩි 30ක්වත් වේගයෙන් ඇවිදීම වැනි ව්‍යායාමයක නිරත වන්න.")
        print("* මාස 6කට වරක්වත් රුධිරයේ සීනි මට්ටම (Fasting Blood Sugar) පරීක්ෂා කරගන්න.")
        
    else:
        print("🚨 අධික අවදානමක් (High Risk)! වහාම ක්‍රියාත්මක වන්න.")
        print("* වහාම වෛද්‍යවරයකු (Doctor) හමුවී සම්පූර්ණ රුධිර පරීක්ෂණයක් සිදුකර උපදෙස් ලබාගන්න.")
        print("* සීනි සහ තෙල් අධික කෑම සම්පූර්ණයෙන්ම පාහේ නතර කරන්න.")
        print("* දෛනික කාර්යබහුලත්වය සහ මානසික ආතතිය (Stress) පාලනය කරගන්න.")

# උදාහරණයක් ලෙස Test set එකේ ඉන්න 5 වැනි රෝගියාගේ අවදානම පරීක්ෂා කරමු
test_patient = X_test[4].reshape(1, -1)
get_diabetes_recommendations(test_patient)
```

```python
from sklearn.model_selection import GridSearchCV
from xgboost import XGBClassifier

# ටෙස්ට් කරන්න ඕන Parameters ලැයිස්තුව (Grid)
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [3, 4, 5],                 # ගසේ ගැඹුර අඩු කිරීමෙන් Overfitting පාලනය වේ
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.7, 0.8, 0.9]
}

xgb = XGBClassifier(random_state=42, eval_metric='logloss')

# GridSearchCV මගින් හොඳම සෙටින්ග්ස් සොයමු
grid_search = GridSearchCV(estimator=xgb, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

# හොඳම Model එක ගනිමු
best_xgb = grid_search.best_estimator_

# අලුත් Accuracy එක බලමු
y_pred_grid = best_xgb.predict(X_test)
grid_acc = accuracy_score(y_test, y_pred_grid)
print(f"Tuned XGBoost Accuracy: {grid_acc * 100:.2f}%")
print("Best Parameters:", grid_search.best_params_)
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.linear_model import LogisticRegression

# 1. SMOTE මගින් Training දත්ත සමතුලිත කිරීම
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

print(f"SMOTE වලට පෙර Class 1 ප්‍රමාණය: {sum(y_train == 1)}")
print(f"SMOTE වලට පසු Class 1 ප්‍රමාණය: {sum(y_train_resampled == 1)}")

# 2. සමතුලිත දත්ත මත අපේ Logistic Regression එක නැවත පුහුණු කිරීම
balanced_log_model = LogisticRegression(random_state=42)
balanced_log_model.fit(X_train_resampled, y_train_resampled)

# 3. අලුත් Accuracy එක බැලීම
balanced_acc = accuracy_score(y_test, balanced_log_model.predict(X_test))
print(f"\nSMOTE කළ පසු Logistic Regression Accuracy: {balanced_acc * 100:.2f}%")
```
